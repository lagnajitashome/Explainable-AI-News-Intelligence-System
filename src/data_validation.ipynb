```python
import sys
import os

sys.path.append("../src")

from data_ingestion import load_data

data = load_data()

print("Data loaded successfully!")
print("Dataset shape:", data.shape)
```

```python
print(data.columns.tolist())
```

```python
required_columns = [
    "title",
    "text",
    "subject",
    "date",
    "label"
]

missing_columns = []

for column in required_columns:
    if column not in data.columns:
        missing_columns.append(column)

print("Missing columns:", missing_columns)
```

```python
if missing_columns:
    print("❌ Missing required columns!")
    print(missing_columns)
else:
    print("✅ All required columns are present!")
```

```python
missing_values = data.isnull().sum()

print("Missing values in each column:\n")
print(missing_values)
```

```python
total_missing = data.isnull().sum().sum()

print("Total missing values:", total_missing)
```

```python
if total_missing > 0:
    print("⚠️ Dataset contains missing values.")
else:
    print("✅ No missing values found!")
    
```

```python
duplicate_rows = data.duplicated().sum()

print("Number of duplicate rows:", duplicate_rows)
```

```python
if duplicate_rows > 0:
    print("⚠️ Duplicate rows found!")
else:
    print("✅ No duplicate rows found!")
```

```python
print("Dataset Data Types:\n")

print(data.dtypes)
```

```python
print("Unique label values:")

print(data["label"].unique())
```

```python
valid_labels = {0, 1}

dataset_labels = set(data["label"].unique())

if dataset_labels.issubset(valid_labels):
    print("✅ All labels are valid!")
else:
    print("❌ Invalid labels found!")
    print("Found labels:", dataset_labels)
```

```python
print("Label Distribution:\n")

print(data["label"].value_counts())
```
