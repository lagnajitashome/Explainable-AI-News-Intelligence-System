```python
import pandas as pd
import numpy as np

```

```python
df = pd.read_csv("../data/processed/preprocessed_news.csv")

df.head()
```

```python
df.columns.tolist()
```

```python
# Create title length feature

df["title_length"] = df["title"].str.len()

df[["title", "title_length"]].head()
```

```python
# Create title word count feature

df["title_word_count"] = df["title"].str.split().str.len()

df[["title", "title_length", "title_word_count"]].head()
```

```python
# Create clean content length feature

df["clean_content_length"] = df["clean_content"].str.len()

df[["clean_content", "clean_content_length"]].head()
```

```python
# Create clean content word count feature

df["clean_word_count"] = df["clean_content"].str.split().str.len()

df[["clean_content", "clean_content_length", "clean_word_count"]].head()
```

```python
# Create title to content length ratio

df["title_content_ratio"] = (
    df["title_length"] /
    (df["clean_content_length"] + 1)
)

df[
    [
        "title_length",
        "clean_content_length",
        "title_content_ratio"
    ]
].head()
```

```python
import string

# Create punctuation count feature

df["punctuation_count"] = df["content"].apply(
    lambda x: sum(char in string.punctuation for char in x)
)

df[["content", "punctuation_count"]].head()
```

```python
# Create number count feature

df["number_count"] = df["content"].apply(
    lambda x: sum(char.isdigit() for char in x)
)

df[["content", "number_count"]].head()
```

```python
# Create uppercase character count feature

df["uppercase_count"] = df["content"].apply(
    lambda x: sum(char.isupper() for char in x)
)

df[["content", "uppercase_count"]].head()
```

```python
# Create uppercase ratio feature

df["uppercase_ratio"] = (
    df["uppercase_count"] /
    (df["text_length"] + 1)
)

df[["text_length", "uppercase_count", "uppercase_ratio"]].head()
```

```python
# Create uppercase ratio feature

df["uppercase_ratio"] = (
    df["uppercase_count"] /
    (df["text_length"] + 1)
)

df[["text_length", "uppercase_count", "uppercase_ratio"]].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# Initialize TF-IDF Vectorizer

tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words="english",
    ngram_range=(1, 2)
)
```

```python
# Check for missing values in clean_content
print("Missing values:", df["clean_content"].isnull().sum())

# Ensure all values are strings
df["clean_content"] = df["clean_content"].fillna("").astype(str)

# Check empty rows
print("Empty rows:", (df["clean_content"].str.strip() == "").sum())
```

```python
# Apply TF-IDF transformation

tfidf_matrix = tfidf.fit_transform(df["clean_content"])

print("TF-IDF transformation completed!")
print("Shape of TF-IDF matrix:", tfidf_matrix.shape)
```

```python
# Remove rows with empty clean_content

df = df[df["clean_content"].str.strip() != ""].copy()

print("Dataset shape after removing empty rows:")
print(df.shape)
```

```python
# Apply TF-IDF again after removing empty rows

tfidf_matrix = tfidf.fit_transform(df["clean_content"])

print("Final TF-IDF matrix shape:", tfidf_matrix.shape)
```

```python
import os

# Create processed folder if it doesn't exist
os.makedirs("../data/processed", exist_ok=True)

# Save feature-engineered dataset
df.to_csv(
    "../data/processed/feature_engineered_news.csv",
    index=False
)

print("Feature-engineered dataset saved successfully!")
print("Location: ../data/processed/feature_engineered_news.csv")
```

```python
from scipy.sparse import save_npz

# Save TF-IDF matrix
save_npz(
    "../data/processed/tfidf_features.npz",
    tfidf_matrix
)

print("TF-IDF matrix saved successfully!")
```
